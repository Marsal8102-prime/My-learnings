This exercise practises using Python conditional statements (`if`, `elif`, and `else`).

## Problem

Given an integer `n`, print:

- `Weird` if `n` is odd.
- `Not Weird` if `n` is even and between 2 and 5 (inclusive).
- `Weird` if `n` is even and between 6 and 20 (inclusive).
- `Not Weird` if `n` is even and greater than 20.

## Approach

1. Check whether the number is odd using `n % 2 != 0`.
2. If it is odd, print `Weird`.
3. Otherwise, check the applicable even-number range and print the required result.

## Concepts Used

- Modulus operator: `%`
- Conditional statements: `if`, `elif`, `else`
- Logical comparisons and inclusive ranges

## Example

| Input | Output |
| --- | --- |
| `3` | `Weird` |
| `4` | `Not Weird` |
| `18` | `Weird` |
| `22` | `Not Weird` |

Source: [HackerRank - Python If-Else](https://www.hackerrank.com/challenges/py-if-else/problem)


```python
def main():
    n = int(input())
    if n>=1 and n<=100:
        if n%2!=0:
            print("Weird")
        elif n%2==0 and n in range(2,5+1):
            print("Not Weird")
        elif n%2==0 and n in range(6,20+1):
            print("Weird")
        else:
            print("Not Weird")


main()
```
